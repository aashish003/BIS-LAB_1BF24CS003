```python
import random

# Feature names and their measurement costs
features = ["F1", "F2", "F3", "F4", "F5", "F6"]
cost = [2, 5, 1, 6, 3, 4]

# Accuracy for different feature subsets is simulated
# for simple lab demonstration
def fitness(chromosome):

    selected = [i for i in range(6) if chromosome[i]]

    if not selected:
        return 0

    # Example classification accuracy
    accuracy = 70 + 5 * len(selected)

    # Total measurement cost
    total_cost = sum(cost[i] for i in selected)

    # Fitness = accuracy - cost penalty
    fitness = accuracy - total_cost

    return fitness


# Initial population
population = [
    [random.randint(0, 1) for _ in range(6)]
    for _ in range(6)
]

# Genetic Algorithm
for generation in range(5):

    scores = [fitness(c) for c in population]

    # Select best two
    best = sorted(
        population,
        key=fitness,
        reverse=True
    )[:2]

    # Crossover
    new_population = best.copy()

    while len(new_population) < 6:

        p1, p2 = random.sample(best, 2)

        point = random.randint(1, 5)

        child = p1[:point] + p2[point:]

        # Mutation
        i = random.randint(0, 5)
        child[i] = 1 - child[i]

        new_population.append(child)

    population = new_population

    print(
        "Generation",
        generation + 1,
        "Best Fitness:",
        fitness(best[0])
    )


# Final solution
best = max(population, key=fitness)

selected = [
    features[i]
    for i in range(6)
    if best[i]
]

total_cost = sum(
    cost[i]
    for i in range(6)
    if best[i]
)

print("\nBest Feature Subset:", selected)
print("Feature Cost:", total_cost)
print("Fitness:", fitness(best))
```
